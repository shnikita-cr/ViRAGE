```python
from pathlib import Path

from rich import print as rprint

from src.visrag_core import VisRAGRequest, VisRAGDataProfile, VisRAGColumnProfile, VisRAGConfig, VisRAGCoreService
```

```python
corpus = Path("../../rag_corpus/data/visrag_corpora/").resolve()
```

```python
request = VisRAGRequest(
    query="show average profit by state",
    data_profile=VisRAGDataProfile(
        columns=[VisRAGColumnProfile(name="State", semantic_type="object", role="dimension")]),
    preferred_chart_types=["bar"],
    selected_fields=["State"],
    top_k=3,
)
```

```python
result = VisRAGCoreService(VisRAGConfig(corpus_root=corpus,
                                        retriever_backend="ollama"
                                        )).search(request)
```
