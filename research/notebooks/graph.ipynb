```python
from __future__ import annotations

from pathlib import Path

from src.application.settings import ViRAGESettings
from src.graph.builder import build_pipeline_graph
from src.infrastructure import RuntimeContext


def export_graph(output_dir: str = "artifacts/langgraph") -> None:
    out = Path(output_dir)
    out.mkdir(parents=True, exist_ok=True)
    settings = ViRAGESettings(artifact_root="artifacts")
    app = build_pipeline_graph(runtime=RuntimeContext(settings=settings))

    # LangGraph compiled graph
    graph = app.get_graph(xray=True)

    # 1. Mermaid text
    mermaid = graph.draw_mermaid()
    (out / "pipeline_graph.mmd").write_text(mermaid, encoding="utf-8")

    # # 2. PNG image
    # png_bytes = graph.draw_mermaid_png()
    # (out / "pipeline_graph.png").write_bytes(png_bytes)

    print(f"Saved Mermaid: {out / 'pipeline_graph.mmd'}")
    # print(f"Saved PNG:     {out / 'pipeline_graph.png'}")

    print(graph.draw_ascii())


if __name__ == "__main__":
    export_graph()
```
