```python
from __future__ import annotations

import sys
from pathlib import Path

ROOT = Path(".").resolve().parents[1]
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.application.contracts import PipelineRequest
from src.application.pipeline import ViRAGEPipeline
from src.application.settings import ViRAGESettings
from rich import print as rprint
```

```python
tmp_path = Path("../demo_data").resolve()
```

```python
from langchain_ollama import ChatOllama

# LLM_MODEL = "gemma3:27b-cloud"  # или "llama3.2:1b"
LLM_MODEL = "gemma4:31b-cloud"  # или "llama3.2:1b"
llm = ChatOllama(
    model=LLM_MODEL,
    temperature=0,
)
```

```python
data_path = (tmp_path / "iris.csv").resolve()
pipeline = ViRAGEPipeline(settings=ViRAGESettings(artifact_root=tmp_path / "artifacts"),
                          reasoning_llm=llm,
                          spec_llm=llm,
                          vlm=llm,
                          vision_judge_llm=llm
                          )
result = pipeline.invoke(
    PipelineRequest(query="Show the distribution of SepalLengthCm for species Iris-setosa in histogram",
                    data_path=data_path.as_posix()))
```
